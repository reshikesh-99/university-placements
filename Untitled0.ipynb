```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("placements.csv")

df['placed'] = df['placed'].apply(lambda x: 1 if str(x).strip().lower() in ['yes', '1', 'true'] else 0)

lower = df['ctc_lpa'].quantile(0.01)
upper = df['ctc_lpa'].quantile(0.99)
df['ctc_lpa'] = np.clip(df['ctc_lpa'], lower, upper)

placement_rate = df.groupby('dept')['placed'].mean().reset_index()
placement_rate.rename(columns={'placed': 'placement_rate'}, inplace=True)
print(placement_rate)

bins = [0, 6, 7, 8, 9, 10]
labels = ['<6', '6-7', '7-8', '8-9', '9-10']
df['cgpa_bucket'] = pd.cut(df['cgpa'], bins=bins, labels=labels, include_lowest=True)

cgpa_analysis = df.groupby('cgpa_bucket').agg(
    placement_rate=('placed', 'mean'),
    median_ctc=('ctc_lpa', 'median')
).reset_index()
print(cgpa_analysis)

df['internships'] = df['internships'].fillna(0)
df['certifications'] = df['certifications'].fillna(0)
print(df[['internships', 'certifications']].isnull().sum())

corr_vars = ['internships', 'certifications', 'cgpa', 'ctc_lpa']
corr_matrix = df[corr_vars].corr()
print(corr_matrix)
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation between Internships, Certifications, CGPA and CTC")
plt.show()

plt.figure(figsize=(8,5))
sns.barplot(data=placement_rate, x='dept', y='placement_rate', palette='viridis')
plt.title('Department-wise Placement Rate')
plt.ylabel('Placement Rate')
plt.xlabel('Department')
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='cgpa', y='ctc_lpa', hue='dept', style='placed', alpha=0.7)
plt.title('CGPA vs CTC (Colored by Department)')
plt.xlabel('CGPA')
plt.ylabel('CTC (LPA)')
plt.show()
```
